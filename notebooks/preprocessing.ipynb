```python
import pandas as pd
import os

folder_path = "../dataset"

files = [f for f in os.listdir(folder_path) if f.endswith(".parquet")]

print("Files found:", len(files))
```

```python
df_list = []

for file in files:
    file_path = os.path.join(folder_path, file)
    print(f"Reading {file}...")
    
    temp_df = pd.read_parquet(file_path)
    df_list.append(temp_df)

merged_df = pd.concat(df_list, ignore_index=True)

print("Merged successfully!")
print("Shape:", merged_df.shape)
```

```python
merged_df.to_csv("../dataset/merged_dataset.csv", index=False)

print("Saved successfully!")
```

```python
df = merged_df  # if not already assigned

print("Shape:", df.shape)
```

```python
print(df.columns)
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df['Label'].value_counts(normalize=True) * 100
```

```python
df['Label'] = df['Label'].apply(lambda x: 0 if x == 'Benign' else 1)

df['Label'].value_counts()
```

```python
cols_to_drop = ['Timestamp', 'Flow ID', 'Src IP', 'Dst IP']

df.drop(columns=[col for col in cols_to_drop if col in df.columns], inplace=True)

print("Remaining columns:", len(df.columns))
```

```python
import numpy as np

df.replace([np.inf, -np.inf], np.nan, inplace=True)
```

```python
df.dropna(inplace=True)
```

```python
X = df.drop('Label', axis=1)
y = df['Label']
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.ensemble import IsolationForest
```

```python
model = IsolationForest(
    n_estimators=100,     # number of trees
    contamination=0.2,    # expected anomaly ratio (~20%)
    random_state=42
)
```

```python
model.fit(X_scaled)
```

```python
y_pred = model.predict(X_scaled)
```

```python
y_pred = [1 if x == -1 else 0 for x in y_pred]
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
print("Accuracy:", accuracy_score(y, y_pred))
print("Classification Report:")
print(classification_report(y, y_pred))
print("Confusion Matrix:")
print(confusion_matrix(y, y_pred))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

```python
model.fit(X_train)
```

```python
y_pred = model.predict(X_test)
y_pred = [1 if x == -1 else 0 for x in y_pred]
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1   # use all CPU cores (faster)
)
```

```python
rf.fit(X_train, y_train)
```

```python
y_pred_rf = rf.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print("\nClassification Report:\n", classification_report(y_test, y_pred_rf))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_rf))
```

```python
#trying to reduce false negatives
rf_bal = RandomForestClassifier(
    n_estimators=100,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)

rf_bal.fit(X_train, y_train)
```

```python
y_prob = rf_bal.predict_proba(X_test)[:, 1]
y_pred_new = [1 if p > 0.3 else 0 for p in y_prob]
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(y_test, y_pred_new))
print(confusion_matrix(y_test, y_pred_new))
```

```python
#feature importance
import pandas as pd

importances = rf_bal.feature_importances_
feature_names = X.columns

feat_imp = pd.Series(importances, index=feature_names)
feat_imp = feat_imp.sort_values(ascending=False)

feat_imp.head(10)
```

```python
#rducing noise and retraining model with top features
top_features = feat_imp.head(15).index
top_features
```

```python
X_df = pd.DataFrame(X_scaled, columns=X.columns)
```

```python
from sklearn.model_selection import train_test_split

X_train_df, X_test_df, y_train, y_test = train_test_split(
    X_df, y, test_size=0.2, random_state=42
)
```

```python
X_train_reduced = X_train_df[top_features]
X_test_reduced = X_test_df[top_features]
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_reduced = RandomForestClassifier(
    n_estimators=100,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)
```

```python
rf_reduced.fit(X_train_reduced, y_train)
```

```python
y_prob_red = rf_reduced.predict_proba(X_test_reduced)[:, 1]
y_pred_red = [1 if p > 0.3 else 0 for p in y_prob_red]
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(y_test, y_pred_red))
print(confusion_matrix(y_test, y_pred_red))
```
